```python
# ── 1. Simple ATM Calculator ──────────────────────────────────────────
balance = float(input("Enter account balance: "))
withdrawal = float(input("Enter withdrawal amount: "))

if withdrawal <= balance:
    balance -= withdrawal
    print(f"Transaction successful. Remaining balance: ₹{balance:.0f}")
else:
    print("Insufficient balance. Transaction failed.")

```

```python
# ── 2. Electricity Bill Generator ────────────────────────────────────
units = int(input("Enter units consumed: "))

if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = (100 * 5) + ((units - 100) * 7)
else:
    bill = (100 * 5) + (100 * 7) + ((units - 200) * 10)

print(f"Bill amount = ₹{bill}")
```

```python
# ── 3. Temperature Converter ──────────────────────────────────────────
temp = float(input("Enter temperature value: "))
unit = input("Enter unit (C or F): ").strip().upper()

if unit == "C":
    converted = (temp * 9/5) + 32
    print(f"Temperature in Fahrenheit: {converted}°F")
elif unit == "F":
    converted = (temp - 32) * 5/9
    print(f"Temperature in Celsius: {converted:.1f}°C")
else:
    print("Invalid unit. Please enter C or F.")
```

```python
# ── 4. Mobile Recharge Validity ───────────────────────────────────────
from datetime import datetime

recharge_input = input("Enter recharge date (YYYY-MM-DD): ")
today_input    = input("Enter today's date  (YYYY-MM-DD): ")

recharge_date = datetime.strptime(recharge_input, "%Y-%m-%d")
today_date    = datetime.strptime(today_input,    "%Y-%m-%d")

days_used = (today_date - recharge_date).days

if days_used <= 28:
    print(f"Plan is still valid. {28 - days_used} day(s) remaining.")
else:
    print("Plan expired. Recharge again.")
```

```python

# ── 5. Bus Seat Reservation System ───────────────────────────────────
booked   = ['Amit', 'Sara', 'John']
cancel   = 'Sara'
new_name = 'Neha'
position = 1

if cancel in booked:
    booked.remove(cancel)
    print(f"'{cancel}' removed from bookings.")
else:
    print(f"'{cancel}' not found in bookings.")

booked.insert(position, new_name)
print(f"Updated booking list: {booked}")
```

```python
# ── 6. Sum of Even Numbers ────────────────────────────────────────────
n = int(input("Enter a number: "))

total = sum(i for i in range(2, n + 1, 2))
print(f"Sum of even numbers: {total}")
```

```python

# ── 7. Dictionary – Student Marks Summary ────────────────────────────
students = {}

for _ in range(3):
    name  = input("Enter name: ")
    marks = float(input("Enter marks: "))
    students[name] = marks

average = sum(students.values()) / len(students)
print(f"Student Marks: {students}")
print(f"Average Marks: {average}")
```

```python

# ── 8a. Dictionary – Word Frequency ──────────────────────────────────
sentence = input("Enter a sentence: ").lower().split()

frequency = {}
for word in sentence:
    frequency[word] = frequency.get(word, 0) + 1

for word, count in frequency.items():
    print(f"{word}: {count}")
```

```python
# ── 8b. Functions – Check Prime ──────────────────────────────────────
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

num = int(input("Enter a number: "))
if is_prime(num):
    print(f"{num} is a prime number.")
else:
    print(f"{num} is not a prime number.")
```

```python
# ── 9. Vowel Counter ──────────────────────────────────────────────────
def count_vowels(text):
    return sum(1 for char in text.lower() if char in "aeiou")

string = input("Enter a string: ")
print(f"Number of vowels: {count_vowels(string)}")
```

```python
# ── 10. Count Divisible Numbers ───────────────────────────────────────
N = int(input("Enter the range limit (N): "))
D = int(input("Enter the divisor (D): "))

count = sum(1 for i in range(1, N + 1) if i % D == 0)
print(f"Count of numbers divisible by {D}: {count}")
```
